```python
import sys
sys.path.insert(0, 'backend')
from app import app

client = app.test_client()
print('Health:', client.get('/api/health').json)

payload = {
    'study_hours': 3,
    'attendance': 85,
    'previous_score': 70,
    'sleep_hours': 8
}
response = client.post('/api/predict', json=payload)
print('Status:', response.status_code)
print('Prediction:', response.json)
assert response.status_code == 200
assert 0 <= response.json['predicted_score'] <= 100
```

```python
# Check validation: invalid attendance should return HTTP 400
bad = {**payload, 'attendance': 150}
response = client.post('/api/predict', json=bad)
print('Invalid input status:', response.status_code, response.json)
assert response.status_code == 400
```
